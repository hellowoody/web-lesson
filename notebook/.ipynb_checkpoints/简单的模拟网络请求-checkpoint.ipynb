```python
import requests
```

```python
def fetch_url(url):
    try:
        response = requests.get(url) 
        if response.status_code == 200:
            print(response.text)
        else:
            print("请求失败：",response.status_code)
    except requests.exceptions.RequestException as e:
        print("发生错误",e)
        
```

```python
url = "https://baidu.com"
fetch_url(url)
```
